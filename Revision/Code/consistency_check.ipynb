```python
import os
import pickle

model_dir = "/Users/adithyamadduri/Desktop/Projects/proteomics_LGBM(ANML+Meta)_fixed"

demo_exact = {"age_at_visit", "educ", "msex"}
demo_prefixes = ("APOE_",)

def is_demographic(f):
    if f in demo_exact:
        return True
    if any(str(f).startswith(p) for p in demo_prefixes):
        return True
    return False

model_files = [f for f in os.listdir(model_dir) if f.endswith("_automl.pkl")]

assert model_files, "No model files found."

found_any = False

for mf in model_files:
    with open(os.path.join(model_dir, mf), "rb") as f:
        automl = pickle.load(f)

    features = automl.model.estimator.feature_name_

    demo_feats = [f for f in features if is_demographic(f)]
    if demo_feats:
        found_any = True
        print(f"\n❌ DEMOGRAPHICS FOUND in {mf}:")
        for d in demo_feats:
            print("   ", d)

if not found_any:
    print("\n✅ NO demographic features appear in ANY model in _fixed.")
```
